```python
#Step 1 Loading libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings(action="ignore")

import os

for dirname, _, filenames in os.walk('/kaggle/input'):
    for filename in filenames:
        print(os.path.join(dirname, filename))

#Step 2 Reading a file
df = pd.read_csv('/kaggle/input/employers-data/Employers_dataset.csv')

#Step 3 Analysing data
print(df.head())
print(df.tail())
print(df.info())
print(df.columns)
print("\nData Types:")
print(df.dtypes)

print("\nSummary Statistics:")
print(df.describe())

print("\nShape:", df.shape)

print(len(df))                                                      # How many employees are there?

# count_department=df.Department.value_counts()                              
# print(count_department)                                             # How many departments exist?

# numerical_data=df.select_dtypes(include="number").columns           
# print(numerical_data)                                               # Which columns are numerical?

# categorical_data=df.select_dtypes(include="object").columns           
# print(categorical_data)                                             # Which columns are categorical?

# print("Employee_ID" in df.columns)                                  # Which column is the employee identifier?                      

#Step 4 Investigate Impossible values           

##Check minimum and maximum values
# print(df.Age.max())
# print(df.Age.min())       

# #Filter impossible values

# print(df[df['Age']>100])
# print(df['Age'].between(0,20).sum())   

#Step 4 Finding missing value

print("\nMissing Values:")
print(df.isnull().sum())

# #Step 5 Finding duplicates
print("\nDuplicate Rows:", df.duplicated().sum())

#Step 6 Finding unique values
# print("\nUnique Values:")
# print(df.nunique())

#Step 7 Remove rows with many missing values
# df=df.dropna()
# print(len(df))

#Step 8 Handle missing value Fill or remove them based on the column:

df["Age"]=df["Age"].fillna(df["Age"].median)
df["Gender"]=df["Gender"].fillna(df["Gender"].mode()[0])
df["Job_Title"]=df["Job_Title"].fillna(df["Job_Title"].mode()[0])
df["Location"]=df["Location"].fillna(df["Location"].mode()[0])
# print(df.head(50))

df["Age"]=pd.to_numeric(df["Age"],errors='coerce')              #errors='coerce' means Invalid values are replaced with NaN.
print(df.head())

# df["Age"]=pd.to_numeric(df["Age"], errors="raise")              #errors="raise" means it raises a ValueError
# print(df.head())

#Before we do EDA, lets separate Numerical and categorical variables for easy analysis

num_col=df.select_dtypes(include=['number']).columns
cat_col=df.select_dtypes(include=['object']).columns
print("Numerical Data:")
print(num_col)

print("Categorical Data:")
print(cat_col)

#Step 9 Univariate Analysis
exclude = ["Employee_ID"]

for col in num_col:
   if col in exclude:
        continue
   print(col)
   print('Skew :', round(df[col].skew(), 2))

   plt.figure(figsize = (15, 4))

# Activate the first plot in a 1×3 layout
   plt.subplot(1, 3, 1)
# Draw histogram for the current numerical column
   sns.histplot(data=df, x=col, kde=True,color="hotpink")
   plt.ylabel("Count")
   plt.xlabel(col)
   plt.title(f"Histogram of {col}")

# Activate the second plot in a 1×3 layout
   plt.subplot(1, 3, 2)
# Draw boxplot Visualize the distribution and detect outliers
   sns.boxplot(data=df,x=col,color="orange")
   plt.ylabel("Count")
   plt.xlabel(col)
   plt.title(f"Boxplot of {col}")

# Activate the third plot in a 1×3 layout
   plt.subplot(1, 3, 3) 
# Draw boxplot Visualize the distribution and detect outliers
   sns.violinplot(data=df,x=col)
   plt.ylabel("Count")
   plt.xlabel(col)
   plt.title(f"violinplot of {col}")

   plt.show()

#Categorical variables are being visualized using a count plot.
#Categorical variables provide the pattern of factors influencing Employee Salary


fig, axes = plt.subplots(3, 2, figsize = (18, 18))
fig.suptitle('Bar plot for all categorical variables in the dataset')

sns.countplot(data=df,x="Gender",color="hotpink",ax=axes[0,0],order=df["Gender"].value_counts().index)
sns.countplot(data=df,x="Department",color="blue",ax=axes[0,1],order=df["Department"].value_counts().index)
sns.countplot(data=df,x="Job_Title",color="orange",ax=axes[1,0],order=df["Job_Title"].value_counts().index)
sns.countplot(data=df,x="Education_Level",color="yellow",ax=axes[1,1],order=df["Education_Level"].value_counts().index)
sns.countplot(data=df,x="Location",color="purple",ax=axes[2,0],order=df["Location"].value_counts().index)
axes[2,1].set_visible(False)
plt.show()

# Step 10:Bivariant Data Analysis
#1. Pairplot for examining the relationship among all numerical data
plt.figure(figsize=(10,8))
sns.pairplot(data=df)
plt.suptitle("Pairplot of Numerical data")
plt.tight_layout()
plt.show()


#2. Violin Plot for examining the relationship between Department and Salary .
df['Department'] = df['Department'].astype(str)
plt.figure(figsize=(10, 8))
sns.violinplot(x="Department", y="Salary", data=df,
               palette={'Engineering': 'lightcoral', 'Sales': 'lightblue', 'Finance': 'lightgreen', 'Marketing': 'gold', 'Product': 'lightskyblue', 'HR': 'lightpink'}, alpha=0.7)
plt.suptitle("Relation between Department and Salary")
plt.show()

#2. BarPlot for examining the relationship between Department and Salary,Genderwise.
df['Department'] = df['Department'].astype(str)
plt.figure(figsize=(10, 8))
sns.barplot(x="Department", y="Salary", data=df,hue="Gender",palette={"Male": "skyblue", "Female": "hotpink"},alpha=0.7)
              #  palette={'Engineering': 'lightcoral', 'Sales': 'lightblue', 'Finance': 'lightgreen', 'Marketing': 'gold', 'Product': 'lightskyblue', 'HR': 'lightpink'}, alpha=0.7)
plt.suptitle("Relation between Department and Salary(Genderwise)")
plt.show()
```
